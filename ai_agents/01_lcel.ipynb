## 기본적인 체인 구성

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
    model = "gpt-5-nano",
    temperature=0,
    # max_token = None
    # timeoht = None
    # max_retries = 3
    # api_key = None
)
```

```python
prompt_template = PromptTemplate.from_template("{topic}에 관련된 농담을 말해줘")
output_parser = StrOutputParser()

#체인 구성
chain = prompt_template | llm | output_parser
```

```python
prompt_template.invoke({"topic":"바다"})
```

```python
response = chain.invoke(
    {"topic":"바다"}
)
```

```python
print(response)
```

#복잡한 체인

```python
from langchain.chat_models import init_chat_model

llm = init_chat_model(
    model = "gpt-5-nano",
    temperature = 0
)

story_prompt = PromptTemplate.from_template("{topic}에 대한 짧은 이야기를 작성해주세요.")
story_chain = story_prompt | llm | StrOutputParser()

analysis_prompt = PromptTemplate.from_template(
    """다음 이야기의 전체적인 분위기를 분석해 주세요

    {story}
    """
)

analysis_chain = analysis_prompt | llm | StrOutputParser()

story_with_analysis_chain = story_chain | analysis_prompt
```

```python
story_with_analysis_chain.invoke(
    {"topic":"바다"}
)
```

## 디버그 Runnable

```python
from langchain_core.runnables import RunnablePassthrough
class DebugPassThrough(RunnablePassthrough):
    def invoke(self, *arg, **kwargs):
        output = super().invoke(*arg,**kwargs)
        print('===== DEBUG OUTPUT =====')
        print(output)
        print("========================")
        return output
```

```python
story_with_analysis_chain = story_chain | DebugPassThrough() | analysis_prompt
```

```python
story_with_analysis_chain.invoke(
    {"topic":"바다"}
)
```

##stream()

```python
response = story_with_analysis_chain.stream(
    {"topic":"바다"}
)

for chunk in response:
    print(chunk, end = "|", flush = True)
```
