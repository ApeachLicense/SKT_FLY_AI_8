```python
import torch
import torch.nn as nn
import torch.optim as optim

X = torch.FloatTensor([[0, 0], [0, 1], [1, 0], [1, 1]])
y = torch.FloatTensor([[0], [1], [1], [0]])
```

```python
model = nn.Sequential(
    nn.Linear(2, 10),
    nn.ReLU(),
    nn.Linear(10,1),
    nn.Sigmoid()
)

criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr = 0.1)
```

```python
model_2 = nn.Sequential(
    nn.Linear(2,10),
    nn.Sigmoid(),
    nn.Linear(10,1),
    nn.Sigmoid()
)
```

```python
def train_model(_model, _x,_y,_optimizer):
  for epoch in range(10000):
    hypothesis = _model(_x)
    loss = criterion(hypothesis, _y)

    _optimizer.zero_grad()
    loss.backward()
    _optimizer.step()

    if epoch % 1000 == 0:
        print(f'Epoch {epoch} Loss: {loss.item():.6f}')
```

```python
def prain_result(_model,_x):
  with torch.no_grad():
   predicted = _model(_x)
   predicted_cls = predicted.round()
   print("\n--- 학습 결과 ---")
   print(f"예측값:\n{predicted.detach().numpy()}")
   print(f"정답:\n{y.numpy()}")
```

```python
train_model(model,X,y,optimizer)
prain_result(model,X)
```

```python
opti = optim.SGD(model_2.parameters(), lr = 0.1)
train_model(model_2,X,y,opti)
prain_result(model_2,X)
```
