```python
df = df.drop('no', axis=1)
```

```python
df.shape
```

```python
from sklearn.model_selection import train_test_split
train, test = train_test_split(df, test_size = 0.2, random_state = SEED)
train.shape, test.shape
```

```python
x_train = train.drop('y',axis = 1)
y_train = train['y']
```

```python
x_mean = x_train.mean()
x_std = x_train.std()

y_mean = y_train.mean()
y_std = y_train.std()
```

```python
x_train_N = (x_train - x_mean) / x_std
y_train_N = (y_train - y_mean) / y_std
```

```python
x_train_N.shape, y_train_N.shape
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(x_train_N,y_train_N)
```

```python
x_test = test.drop('y',axis=1)
y_test = test['y']
```

```python
x_test_N = (x_test - x_mean) / x_std
y_test_N = (y_test - y_mean) / y_std
```

```python
y_pred = model.predict(x_test_N)
y_pred = y_pred * y_std + y_mean
```

```python
from sklearn.metrics import mean_squared_error, r2_score

mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
print(f"MSE :{mse}, R2 : {r2}")
```

```python
from sklearn.linear_model import Ridge
ridge = Ridge()
from sklearn.linear_model import Lasso
lasso = Lasso()
```

```python
ridge.fit(x_train_N,y_train_N)
y_pred_ridge = ridge.predict(x_test_N)
y_pred_ridge = y_pred_ridge * y_std + y_mean
mse = mean_squared_error(y_test, y_pred_ridge)
r2 = r2_score(y_test, y_pred_ridge)
print(f"MSE :{mse}, R2 : {r2}")
```

```python
lasso.fit(x_train_N,y_train_N)
y_pred_lasso = lasso.predict(x_test_N)
y_pred_lasso = y_pred_lasso * y_std + y_mean
mse = mean_squared_error(y_test, y_pred_lasso)
r2 = r2_score(y_test, y_pred_lasso)
print(f"MSE :{mse}, R2 : {r2}")
```
