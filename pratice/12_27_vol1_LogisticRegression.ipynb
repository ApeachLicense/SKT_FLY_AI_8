```python
import numpy as np
import matplotlib.pyplot as plt

class LogisticRegression:
  def __init__(self, l_r = 0.01, n_iter= 1000):
    self.lr = l_r
    self.n_iters = n_iter
    self.weights = None
    self.bias = None
    self.loses = []

  def _sigmoid(self, z):
    return 1/ (1 + np.exp(-z))

  def _binary_cross_entropy(self, y_true, y_pred):
    epsilon = 1e-9
    term_1 = (y_true) * np.log(y_pred + epsilon)
    term_2 = (1- y_true) * np.log(1-y_pred + epsilon)
    loss = - np.sum(term_1 + term_2) / len(y_true)
    return loss

  def fit(self, X, y):
    n_samples, n_features = X.shape

    self.w = np.zeros(n_features)
    self.b = 0

    for i in range(self.n_iters):
      z = np.dot(X, self.w) + self.b
      y_pred = self._sigmoid(z)

      loss = self._binary_cross_entropy(y, y_pred)
      self.loses.append(loss)

      dw = (1 / n_samples) * np.dot(X.T, (y_pred - y))
      db = (1 / n_samples) * np.sum(y_pred - y)

      self.w -= self.lr * dw
      self.b -= self.lr * db

      if i % 100 == 0:
        print(f"Iteration: {i}, Loss: {loss:.4f}")

  def predict(self, X):
    z = np.dot(X,self.w) + self.b
    y_pred_proba = self._sigmoid(z)
    # Convert probabilities to binary class labels (0 or 1)
    y_pred_class = (y_pred_proba > 0.5).astype(int)
    return y_pred_class
```

```python
X_train = np.array([
        [2, 30], [4, 40], [6, 60], [8, 80],
        [1, 20], [3, 20], [7, 85], [9, 90]
    ])
y_train = np.array([0, 0, 1, 1, 0, 0, 1, 1])
```

```python
model = LogisticRegression(l_r=0.01, n_iter=1000)
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_train)
```
