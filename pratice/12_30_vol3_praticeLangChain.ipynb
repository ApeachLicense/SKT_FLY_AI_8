#LECL (LangChain Expression Languare)

```python
chain = (
    {"object": lambda x: x["var"]}
    | template
    | openai
)
```

```python
response = chain.invoke({"var":"Google 검색엔진"})
```

```python
print(response.content)
```

```python
template = PromptTemplate(
    template="{object}의 창립자는 누구입니까? 실시간 {topic}은 얼마입니까? ",
    input_variables = ["object","topic"]
)
```

```python
chain = (
    {"object": lambda x: x["var1"],"topic": lambda x: x["var2"]}
    | template
    | openai
    | {"result": lambda x: x.content}
)
```

```python
response = chain.invoke({"var1":"nvidia","var2":"주가"})
```

```python
print(response.content)
```

#출력의 구조화

```python
from pydantic import BaseModel, Field
```

```python
from typing import List
```

```python
class Recipe(BaseModel):
  title: str = Field(description="레시피의 제목입니다.")
  ingredients: str = Field(description="요리를 만들기 위한 재료 리스트입니다.")
  steps: List[str] = Field(description="요리 순서 리스트입니다.")
```

```python
struct_return = openai.with_structured_output(Recipe)
```

```python
template = PromptTemplate(
    template = "당신은 뛰어난 요리사입니다. 다음 {food}에 대한 레시프를 만들어주세요.",
    input_variables = ["food"]
)
```

```python
chain = (
    {"food": lambda x: x["Food"]} # 'object' 대신 'food'로 변경하여 PromptTemplate의 input_variables와 일치시킵니다.
    | template
    | struct_return
)
```

```python
response = chain.invoke({"Food":"피자"})
```

```python
dict(response)
```

```python
from langchain_chroma import Chroma

chroma = Chroma.from_documents(
    documents = chunks,
    embedding = embeddings,
    persist_directory = 'db',
    collection_name = 'database'
)
```

```python
retriever = chroma.as_retriever(
    search_type = 'similarity',
    search_kwargs ={"k":3}
)
```

```python
from langchain_openai import ChatOpenAI

model = ChatOpenAI(
    temperature=0.1,
    model_name='gpt-5-nano',
    openai_api_key=api,
)
```

```python
from langchain_core.prompts import PromptTemplate
```

```python
prompt = """You are an assistant for question-answering tasks. Use the following pieces of retrieved context to answer the question. If you don't know the answer, just say that you don't know. Use three sentences maximum and keep the answer concise.
Question: {question}
Context: {context}
Answer:"""
```

```python
prompt_template = PromptTemplate.from_template(prompt)
```

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
chain = (
    {
        "context" : lambda x: x["context"],
        "question" : lambda x: x["question"]
    }
    | prompt_template
    | model
    | StrOutputParser()
)
```

```python
question = "김진욱은 누구인가?"
context = retriever.invoke(question) # "quetion" 대신 question 변수를 전달하도록 수정
chain.invoke({"question":question,"context":context})
```

```python
question = "김진욱이 전공공부를 위해 들인 노력에 대해서 설명해줘"
context = retriever.invoke(question)
chain.invoke({"question":question,"context":context})
```

```python
question = "GitHub Copilot이 그의 레포지토리를 어떻게 평가했는지 설명해줘"
context = retriever.invoke(question)
chain.invoke({"question":question,"context":context})
```
